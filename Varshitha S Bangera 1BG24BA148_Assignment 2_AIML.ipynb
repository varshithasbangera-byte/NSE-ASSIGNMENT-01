```python
print("\nMissing values:\n", df.isnull().sum())
```

```python
print("\nDuplicate rows:", df.duplicated().sum())
```

```python
df = df.drop_duplicates()
```

```python
if 'Date' in df.columns:
    df['Date'] = pd.to_datetime(df['Date'], errors='coerce')
```

```python
if 'Amount' in df.columns:
    df['Amount'] = (df['Amount']
                    .astype(str)
                    .str.replace('[^0-9.-]', '', regex=True)
                    .astype(float))
```

```python
amount = df.select_dtypes(include=['float64', 'int64']).columns
```

```python
df[amount] = df[amount].fillna(df[amount].mean())
```

```python
cat_cols = df.select_dtypes(include=['object']).columns
for col in cat_cols:
    df[col] = df[col].fillna(df[col].mode()[0])
```

```python
for col in cat_cols:
    df[col] = df[col].astype(str).str.strip().str.title()
```

```python
for col in amount:
    q1, q3 = df[col].quantile([0.25, 0.75])
    iqr = q3 - q1
    lower, upper = q1 - 1.5 * iqr, q3 + 1.5 * iqr
    df[col] = np.where(df[col] < lower, np.nan, df[col])
    df[col] = np.where(df[col] > upper, np.nan, df[col])
df[amount] = df[amount].fillna(df[amount].median())
```

```python
if 'Amount' in df.columns:
    df['Expense_Level'] = np.where(df['Amount'] > df['Amount'].median(), 'High', 'Low')
```

```python
if 'Date' in df.columns:
    df['Month'] = df['Date'].dt.month_name()

# (c) Create a derived metric – Annualized Spending
if 'Amount' in df.columns:
    df['Annualized_Spending'] = df['Amount'] * 12
```

```python
print("\n INSIGHT 1: Average Amount by Category ")
if 'user_id	' in df.columns:
    print(df.groupby('user_id')['Amount']
          .agg(['mean', 'sum'])
          .sort_values(by='sum', ascending=False)
          .head())
```

```python
print("\n--- INSIGHT 2: Top 5 Highest Expenses ---")
if 'Amount' in df.columns:
    print(df.sort_values('Amount', ascending=False).head(5))
```

```python
print("\n--- INSIGHT 3: Most Common Payment Methods ---")
if 'Payment Method' in df.columns:
    print(df['Payment Method'].value_counts().head())
```

```python
print("\n--- INSIGHT 3: Most Common Payment Methods ---")
if 'Payment Method' in df.columns:
    print(df['Payment Method'].value_counts().head())
```

```python
print("\n--- INSIGHT 4: Monthly Spending by Expense Level ---")
if 'Month' in df.columns and 'Expense_Level' in df.columns:
    print(pd.pivot_table(df, values='Amount', index='Month', columns='Expense_Level', aggfunc='sum'))
```

```python
print("\n--- INSIGHT 5: Transactions Above Average Spending ---")
if 'Amount' in df.columns:
    avg_amt = df['Amount'].mean()
    print(df[df['Amount'] > avg_amt][['Category', 'Amount']].head())
```
