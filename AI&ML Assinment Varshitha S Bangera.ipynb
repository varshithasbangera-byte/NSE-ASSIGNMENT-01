```python
import pandas as pd
df=pd.read_csv(r"C:\Users\user\Downloads\Students Social Media Addiction.csv")
df
```

```python
df.head()
```

```python
students = df.to_dict(orient="records")
```

```python
len(students)
```

```python
def view_students():
    if len(students) == 0:
        print("No student records available.")
    else:
        for i, student in enumerate(students, start=1):
            print(f"\nStudent {i}")
            for key, value in student.items():
                print(key, ":", value)
```

```python
view_students()
```

```python
def search_by_gender():
    gender = input("Enter gender to search: ").strip().title()
    found = False

    for student in students:
        if student.get("Gender") == gender:
            print(student)
            found = True

    if not found:
        print("No matching records found.")
```

```python
search_by_gender()
```

```python
def average_usage():
    total = 0
    count = 0

    for student in students:
        total += student.get("Daily_Usage_Hours", 0)
        count += 1

    if count > 0:
        print("Average Daily Usage Hours:", round(total / count, 2))
    else:
        print("No data available.")
```

```python
average_usage()
```

```python
def save_to_file():
    new_df = pd.DataFrame(students)
    new_df.to_csv("Student_Addiction_Output.csv", index=False)
    print("Data saved successfully to Student_Addiction_Output.csv")
```

```python
save_to_file()
```

```python
while True:
    print("\n--- STUDENT SOCIAL MEDIA ADDICTION SYSTEM ---")
    print("1. View all students")
    print("2. Search students by gender")
    print("3. Show average daily usage")
    print("4. Save records to file")
    print("5. Exit")

    try:
        choice = int(input("Enter your choice: "))

        if choice == 1:
            view_students()

        elif choice == 2:
            search_by_gender()

        elif choice == 3:
            average_usage()

        elif choice == 4:
            save_to_file()

        elif choice == 5:
            print("Program exited.")
            break

        else:
            print("Invalid choice. Enter 1 to 5.")

    except ValueError:
        print("Please enter a valid number.")
```
